```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from diffusers import StableDiffusionPipeline
from gtts import gTTS
import torch

topic = "The benefits of renewable energy"

# --------------------------------------------------
# 1. TEXT GENERATION
# --------------------------------------------------

print("Generating text...")

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text_prompt = f"Write a short, engaging paragraph about {topic}."

inputs = tokenizer(
    text_prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_length=80
)

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("\nGenerated Text:")
print(generated_text)


# --------------------------------------------------
# 2. IMAGE GENERATION
# --------------------------------------------------

print("\nGenerating image...")

device = "cuda" if torch.cuda.is_available() else "cpu"

if device == "cuda":
    dtype = torch.float16
else:
    dtype = torch.float32

print("Using device:", device)

sd_pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=dtype
)

sd_pipe = sd_pipe.to(device)

image_prompt = (
    "A beautiful digital art illustration representing "
    "the benefits of renewable energy, solar panels, "
    "wind turbines, clean environment, green energy, "
    "sustainable future"
)

image = sd_pipe(
    image_prompt,
    num_inference_steps=25
).images[0]

image.save("content_image.png")

print("Image saved as content_image.png")


# --------------------------------------------------
# 3. AUDIO GENERATION
# --------------------------------------------------

print("\nGenerating audio...")

tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("content_audio.mp3")

print("Audio saved as content_audio.mp3")


# --------------------------------------------------
# FINAL
# --------------------------------------------------

print("\nAll content generated successfully!")
```
